```python
import requests
import os
from dotenv import load_dotenv
import pandas as pd
import yfinance as yf
import csv
```

On check si la clé API de la FRED marche.

```python
load_dotenv()
key = os.getenv("FRED_API_KEY")
r = requests.get("https://api.stlouisfed.org/fred/series/observations",
                 params={"series_id": "UNRATE", "api_key": key, "file_type": "json"})
print(r.status_code, r.text[:300])
```

```python
FRED = pd.DataFrame(r.json()["observations"])[["date", "value"]]
FRED["date"] = pd.to_datetime(FRED["date"])
FRED["value"] = pd.to_numeric(FRED["value"], errors="coerce")  #FRED écrit "." pour les valeurs manquantes
FRED = FRED.set_index("date").rename(columns={"value": "UNRATE"})
FRED.tail()
```

Here we are going to get the S&P500 and the VIX with the yfinance library.

```python
SP500 = yf.download("^GSPC", start="2000-01-01", end="2026-11-21", auto_adjust=True) #january 2000 to november 2026 = 322 months
VIX = yf.download("^VIX", start="2000-01-01", end="2026-11-21", auto_adjust=True)
```

```python
SP500.to_csv("../data/raw/SP500.csv", sep = ";")
VIX.to_csv("../data/raw/VIX.csv", sep = ";")
```
